## 🤖 Step 5: Start the Telegram Bot
Run this cell to launch your Quiz Automation Bot!
Once running, go to Telegram, find your bot, and send **`/postquiz`**!

```python
import asyncio
import nest_asyncio
import logging
from telegram import Update, InlineKeyboardButton, InlineKeyboardMarkup, Poll
from telegram.ext import Application, CommandHandler, CallbackQueryHandler, ContextTypes

# Enable nested event loop in Google Colab
nest_asyncio.apply()

logging.basicConfig(format="%(asctime)s - %(levelname)s - %(message)s", level=logging.INFO)

user_sessions = {}

# --- Helper Functions ---
def get_categories():
    docs = db.collection("questions").where("is_used", "==", False).stream()
    cat_counts = {}
    for doc in docs:
        cat = doc.to_dict().get("category", "General")
        cat_counts[cat] = cat_counts.get(cat, 0) + 1
    return cat_counts

def get_unused_questions(category, limit):
    query = db.collection("questions").where("category", "==", category).where("is_used", "==", False).limit(limit)
    questions = []
    for doc in query.stream():
        data = doc.to_dict()
        data["id"] = doc.id
        questions.append(data)
    return questions

def mark_as_used(question_ids):
    batch = db.batch()
    now = datetime.datetime.utcnow()
    for q_id in question_ids:
        ref = db.collection("questions").document(q_id)
        batch.update(ref, {"is_used": True, "posted_at": now})
    batch.commit()

# --- Telegram Command Handlers ---
async def start_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user = update.effective_user
    keyboard = [
        [InlineKeyboardButton("🚀 Post a Quiz Now", callback_data="start_postquiz")],
        [InlineKeyboardButton("📊 View Database Stats", callback_data="view_stats")]
    ]
    await update.message.reply_text(
        f"👋 Hello, **{user.first_name}**!\n\n🎯 **Quiz Automation Bot is Active**\n\nSend `/postquiz` to start!",
        parse_mode="Markdown",
        reply_markup=InlineKeyboardMarkup(keyboard)
    )

async def postquiz_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    cat_counts = get_categories()
    if not cat_counts:
        text = "⚠️ **No unused questions available in Firestore!**\nPlease run Step 4 to upload questions."
        if update.message:
            await update.message.reply_text(text, parse_mode="Markdown")
        elif update.callback_query:
            await update.callback_query.message.edit_text(text, parse_mode="Markdown")
        return

    keyboard = []
    for cat, count in cat_counts.items():
        keyboard.append([InlineKeyboardButton(f"📚 {cat} ({count} available)", callback_data=f"cat:{cat}")])
    keyboard.append([InlineKeyboardButton("❌ Cancel", callback_data="cancel_wizard")])

    prompt = "🎯 **Step 1/3: Select a Quiz Category**\n\nChoose the category you want to post:"
    if update.message:
        await update.message.reply_text(prompt, parse_mode="Markdown", reply_markup=InlineKeyboardMarkup(keyboard))
    elif update.callback_query:
        await update.callback_query.message.edit_text(prompt, parse_mode="Markdown", reply_markup=InlineKeyboardMarkup(keyboard))

async def handle_callback_query(update: Update, context: ContextTypes.DEFAULT_TYPE):
    query = update.callback_query
    await query.answer()
    data = query.data
    user_id = query.from_user.id

    if data == "cancel_wizard":
        user_sessions.pop(user_id, None)
        await query.edit_message_text("❌ Quiz posting cancelled.")
        return

    if data == "start_postquiz":
        await postquiz_command(update, context)
        return

    if data == "view_stats":
        cat_counts = get_categories()
        total_unused = sum(cat_counts.values())
        text = f"📊 **Database Statistics**\n\n• **Total Unused Questions:** `{total_unused}`\n\n"
        for cat, cnt in cat_counts.items():
            text += f"• **{cat}:** `{cnt}` questions\n"
        await query.edit_message_text(text, parse_mode="Markdown")
        return

    # Step 2: Category Selected -> Number of questions
    if data.startswith("cat:"):
        cat = data.split("cat:", 1)[1]
        cat_counts = get_categories()
        available = cat_counts.get(cat, 0)
        if available == 0:
            await query.edit_message_text(f"⚠️ No questions left in **{cat}**.", parse_mode="Markdown")
            return

        user_sessions[user_id] = {"category": cat, "available": available}
        counts = [c for c in [1, 2, 3, 5, 10] if c <= available]
        if available not in counts:
            counts.append(available)
            counts.sort()

        keyboard = []
        row = []
        for c in counts:
            row.append(InlineKeyboardButton(f"{c} Question{'s' if c > 1 else ''}", callback_data=f"cnt:{c}"))
            if len(row) == 2:
                keyboard.append(row)
                row = []
        if row:
            keyboard.append(row)
        keyboard.append([InlineKeyboardButton("🔙 Back", callback_data="start_postquiz")])

        prompt = f"🎯 **Step 2/3: Number of Questions**\n\nCategory: **{cat}** (Available: `{available}`)\n\nHow many questions to post?"
        await query.edit_message_text(prompt, parse_mode="Markdown", reply_markup=InlineKeyboardMarkup(keyboard))
        return

    # Step 3: Question Count Selected -> Schedule timing
    if data.startswith("cnt:"):
        count = int(data.split("cnt:", 1)[1])
        if user_id not in user_sessions:
            await query.edit_message_text("⚠️ Session expired. Type /postquiz again.")
            return
        user_sessions[user_id]["count"] = count
        cat = user_sessions[user_id]["category"]

        keyboard = [
            [InlineKeyboardButton("🚀 Post Immediately (Now)", callback_data="time:0")],
            [InlineKeyboardButton("⏱ In 5 Mins", callback_data="time:5"), InlineKeyboardButton("⏱ In 15 Mins", callback_data="time:15")],
            [InlineKeyboardButton("⏱ In 30 Mins", callback_data="time:30"), InlineKeyboardButton("⏱ In 1 Hour", callback_data="time:60")],
            [InlineKeyboardButton("❌ Cancel", callback_data="cancel_wizard")]
        ]
        prompt = f"🎯 **Step 3/3: Schedule Quiz Posting**\n\n• Category: **{cat}**\n• Questions: **{count}**\n• Destination: `{TELEGRAM_CHANNEL_ID}`\n\nWhen to post?"
        await query.edit_message_text(prompt, parse_mode="Markdown", reply_markup=InlineKeyboardMarkup(keyboard))
        return

    # Step 4: When to post selected -> Execute!
    if data.startswith("time:"):
        delay = int(data.split("time:", 1)[1])
        if user_id not in user_sessions:
            await query.edit_message_text("⚠️ Session expired. Type /postquiz again.")
            return
        session = user_sessions.pop(user_id)
        cat = session["category"]
        count = session["count"]

        if delay == 0:
            await query.edit_message_text(f"⏳ Fetching `{count}` questions from **{cat}** and posting now...", parse_mode="Markdown")
            asyncio.create_task(post_quiz_to_channel(context.bot, query.message.chat_id, cat, count))
        else:
            await query.edit_message_text(
                f"⏰ **Quiz Scheduled!**\n\n• Category: **{cat}**\n• Count: **{count}**\n• In: **{delay} minutes**\n\nI will post automatically! 🚀",
                parse_mode="Markdown"
            )
            asyncio.create_task(delayed_post_task(context.bot, query.message.chat_id, cat, count, delay))

async def delayed_post_task(bot, admin_chat_id, cat, count, delay_minutes):
    await asyncio.sleep(delay_minutes * 60)
    await post_quiz_to_channel(bot, admin_chat_id, cat, count)

async def post_quiz_to_channel(bot, admin_chat_id, cat, count):
    try:
        questions = get_unused_questions(cat, count)
        if not questions:
            await bot.send_message(chat_id=admin_chat_id, text=f"⚠️ No unused questions left for **{cat}**.", parse_mode="Markdown")
            return

        posted_ids = []
        for idx, q in enumerate(questions, start=1):
            q_title = f"[{cat} #{idx}] {q['question_text']}"[:295]
            expl = q.get("explanation", "")
            await bot.send_poll(
                chat_id=TELEGRAM_CHANNEL_ID,
                question=q_title,
                options=q.get("options", []),
                type=Poll.QUIZ,
                correct_option_id=int(q.get("correct_option_id", 0)),
                explanation=expl[:195] if expl else None,
                is_anonymous=True
            )
            posted_ids.append(q["id"])
            await asyncio.sleep(1.5)

        mark_as_used(posted_ids)
        await bot.send_message(
            chat_id=admin_chat_id,
            text=f"✅ **Successfully posted {len(posted_ids)} quiz questions to `{TELEGRAM_CHANNEL_ID}`!**\n\n• Category: **{cat}**\n• Database updated (questions marked used).",
            parse_mode="Markdown"
        )
    except Exception as e:
        await bot.send_message(chat_id=admin_chat_id, text=f"❌ **Error posting quiz:** `{e}`", parse_mode="Markdown")

# Start bot
app = Application.builder().token(TELEGRAM_BOT_TOKEN).build()
app.add_handler(CommandHandler("start", start_command))
app.add_handler(CommandHandler("postquiz", postquiz_command))
app.add_handler(CallbackQueryHandler(handle_callback_query))

print("🚀 Bot is running! Go to Telegram, message your bot, and send /postquiz!")
await app.run_polling()
```
